```python
clasicos = {"Casablanca", "Titanic", "Forrest Gump", "El Padrino", "Gladiator"}
scifi = {"Matrix", "Interestellar", "Inception", "Titanic", "Avatar"}
animacion = {"Up", "Coco", "Toy Story", "Shrek", "Avatar"}
```

```python
# Len
print(len(clasicos))

# Copy -> copia superficial
clasicos_copiados = clasicos.copy()
print(clasicos_copiados)

# add -> añade un nuevo elemento al set
scifi.add("2001: Odisea en el espacio")
scifi.add("Matrix")
print(scifi)

# update -> mediante uina lista añade varios elementos al setç
animacion.update(["El castillo ambulante", "Titan A.E."])
print(animacion)

# pop -> Elimina un elemento aleatorio y no se puede elegir cuál
#clasicos.pop()
#print(clasicos)

# remove -> Elimina un elemento especifico y si no lo encuentra, da un error:
#clasicos.remove("Gladiator")
#print(clasicos)

# discard -> Elimina un elemento especifico y si no lo encuentra no da error:
clasicos.discard("Bracula II")

#clear -> Vacía el set por completo
clasicos.clear()
print(clasicos)
```

```python
# OPERACIONES ENTRE SETS O CONJUNTOS
clasicos = {"Casablanca", "Titanic", "Forrest Gump", "El Padrino", "Gladiator"}
scifi = {"Matrix", "Interestellar", "Inception", "Titanic", "Avatar"}
animacion = {"Up", "Coco", "Toy Story", "Shrek", "Avatar"}

# union -> | une los sets y elimina los duplicados
peliculas_union = clasicos.union(scifi)
# También se puede con el pipe: peliculas_union = clasicos | scifi
print(peliculas_union)

# intersection -> & saca los que se comparten en ambos sets y si no hay devuelve un set vacío
peliculas_comunes = clasicos.intersection(scifi)
# peliculas_comunes = clasicos & scifi
print(peliculas_comunes)

#diference -> - Devuelve la diferencia de un set inicial con otro 
solo_clasicos = clasicos.difference(scifi)
#solo_clasicos = clasicos - scifi
print(solo_clasicos)

# symetric difference -> ^ Indica las peliculas que está en uno y otro conjunto pero no los que están en ambos
peliculas_unicas = clasicos.symmetric_difference(scifi)
print(peliculas_unicas)

#isdisjoint -> Comprueba si no tienen elementos en común los conjuntos (false si hay y true si no)
print(clasicos.isdisjoint(scifi))

#issubset y issuperset -> comprueba si un conjunto está dentro del otro
mini_set = {"Titanic", "El Padrino"}
print(mini_set.issubset(clasicos))
print(clasicos.issuperset(mini_set))
```
