## Diccionarios

```python
hulk = {
    "nombre": "Hulk",
    "identidad": "Bruce Banner",
    "edad": 48,
}
print(hulk)
```

```python
# también se puede crear un diccionario con la función dict
ironman = dict(nombre = "Iron Man", identidad = "Tony Stark", edad = 50)
print(ironman)
```

```python
 # Indexación
ironman = {
    'nombre': 'Iron Man', 
    'identidad': 'Tony Stark', 
    'edad': 50,
}
print(ironman['nombre'])
#Si se intenta entrar con una clave que no existe, el código se rompe
#Existe un método alternativo para acceder a claves de un diccionario sin que arroje error y como
#segundo argumento se define el mensaje a mostrar cuando no exista la clave
print(ironman.get("ciudad", "No se especifica"))
print(ironman.get("nombre", "No se especifica"))
```

```python
#Modificar
ironman["identidad"] = "Daniel Stark"
print(ironman)

#Añadir clave
ironman["ciudad"] = "Elche"
print(ironman)
```

```python
#Funciones o métodos:

#keys -> devuelve una vista con todas las claves
print(ironman.keys())

lista_keys = list(ironman.keys())
print(lista_keys)

#Values -> devuelve una vista con todos los valores:
print(ironman.values)

#items -> devuelve una tupla de clave-valor de cada elemento del diccionario
print(ironman.items())

#update -> recibe por argum,ento otro diccionario y actualiza el diccionario actual
ironman.update({
    "identidad": "Tony Stark",
    "edad": 51,
    "coche": "Seat Ibiza"
})
print(ironman)

# setdefault -> devuelve el valor de una clave si existe y si no existe la crea con un valor por defecto indicado
ironman.setdefault("pais", "Disparos Unidos")
print(ironman)

# pop -> elimina un elemento de un diccionario y devuelve su valor, si se indica un valor que no existe, salta un error a menor de poner un none como segundo argumento
print(ironman.pop("coche"))
print(ironman)

ironman.pop("novia", None)
print(ironman)

# popitem -> Elimina y devuelve el ultimo par de clave y valor insertado
print(ironman.popitem())
print(ironman)

# copy -> Copia sin referencia o superficial del diccionario
ironman_copy = ironman.copy()
ironman_copy ["nombre"] = "Aquaman"
print(ironman)
print(ironman_copy)

# clear -> limpia todo el diccionario
ironman_copy.clear()
print(ironman_copy)
print(ironman)
```
