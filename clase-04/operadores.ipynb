## Operadores

```python
# Operadores de concatenación:
saludo = "Hola, "
nombre = "Daniel"
edad = 27

mensaje = saludo + nombre + " y tengo " + str(edad) + "años"
print(mensaje)
```

```python
# Operadores de comparación
y = 10
x = 20
print(y==x)
print(y!=x)
print(y > x)
print(y >= x)
print(y < x)
print(y <= x)

print(10 =="10")
print(10==10)
```

```python
# Operadores lógicos

tiene_coche = True
tiene_carnet = False

# Tienes que tener coche Y carnet
print(tiene_coche and tiene_carnet)
# Caso ficticio si tienes coche el carnet es opcional
print (tiene_carnet or tiene_coche)
```

```python
# Operaciones aritméticas directas
a = 30
b = 4

resultado = a+b
resultado = resultado+5
resultado+=5
print(resultado)
```

```python
# Conversiones de tipo de dato

#Convertir de String a numero entero:
numero = int("5")
print(numero)

#Convertir de String a decimal:
numero_decimal = float("5.6")
print(numero_decimal)

#Convertir de numero a string
numero_string = str(106)
print(numero_string)

#Convertir de número a boolean
numero_boolean = bool(0)
print(numero_boolean)
```
