## Condicionales

```python
espectadores = 10

if espectadores > 5: 
    print("Abrimos la sala de cine")
else:
    print("No la abrimos")
```

```python
puntuacion = 8

if puntuacion >= 9:
    print("Obra maestra")
elif puntuacion >= 7:
    print("Notable")
elif puntuacion >= 4:
    print("Pasable")
elif puntuacion >=2:
    print("Mala")
else:
    print("Mierdón")
```

```python
dia = "viernes"
hora = 17

if dia == "viernes" and hora >= 15:
    print("Sesión especial")
elif dia == "sábado" or dia == "domingo":
    print("Horario fin de semana")
else:
    print("Horario normal de semana")
```
