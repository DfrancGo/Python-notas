## Strings

```python
palabra = "Trufaloca"
print(palabra)
```

```python
# indexación -> indices
print(palabra[0])
print(palabra[3])

# indexación negativa
print(palabra[-1])
```

```python
#Rangos de indexación
#String[start:stop:step]
print(palabra[0:4])
print(palabra[:4])
print(palabra[2:6:2])
print(palabra[1:])
print(palabra[::2])
print(palabra[::-1])
```

```python
frase = "Trufa es una perra Mestiza"
frase_rota = "   Trufa es una perra Mestiza    "
print(frase.lower()) #Todo a minúsculas
print(frase.upper()) #Todo a mayúsculas
print(frase.capitalize()) #Hace la primera letra en mayúsculas
print(frase_rota)
print(frase_rota.strip()) #Quita los espacios del principio y final
print(frase.replace("perra", "gata")) #Reemplaza la palabra 'perra' por 'gata'
print(frase.find("T"))#Devuelve el indice de la primera coincidencia
print(frase.index("t"))#hace lo mismo que find pero cuando no encuentra el caracter, da un error

#Extra
heroes = ["iron-man", "Thor", "Hawk-eye"]
print(", ".join(heroes)) #Junta todos los elementos de la lista y lo convierte en un string, la primera parte define qué carácter separa cada elemento
```
