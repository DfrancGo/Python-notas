### range
Recibe tres argumentos, start, stop, step (Inicio, final e incrementos)
Esto lo comparte con otros elementos iterables de python como las listas y los strings

```python
for n in range(5):
    print(n)
#Los limites son excluyente por lo que con el cinco sólo mostrará los valores de 0 a 4    

print("--------------")

for n in range(1,6):
    print(n)

print("--------------")

for n in range(0, 10, 2):
    print(n)

print("--------------")

for n in range(5,0,-1):
    print(n)

#Como el bucle for de Java y JS for(i, i<10, i++), en este caso: for(i = 5 , i>0, i--)

print("--------------")

list_numero =  []

for n in range(1,21):
    list_numero.append(n)

print(list_numero)
```

### enumerate
enumera los elementos de algo iterable, usandose en bucles.
Puede recibir como segundo parametro, desde dónde empezar a enumerar

```python
frutas = ["melocotón", "fresa", "papaya", "kiwi", "melón"]

for indice, fruta in enumerate(frutas):
    print(f"{fruta} - {indice}")

print("--------------")

for indice, fruta in enumerate(frutas, start=1):
    print(f"{fruta} - {indice}")
```

### zip
Combina listas en paralelo y devuelve tuplas emparejando las posiciones de las listas, si alguna de las dos listas no es igual en tamaño, lo que no se cruce no lo devuelve

```python
alumnos = ["Daniel", "Ezequiel", "Fermín", "Paco", "Pepe", "Alejandro", "Inés", "Sandra"]

notas = [5,6,7,3,8,9,2,7]

for alumno, nota in zip (alumnos, notas):
    print(f"{alumno} : {nota}")

boletin = dict(zip(alumnos, notas))   
print(boletin) 
```
