## Listas

```python
heroes = ["Spider Man", "Iron Man", "Hulk"]

#Función len
print(len(heroes))
print(heroes[::-1]) #Lo imprime con las posiciones invertidas
```

```python
universo_marvel = [
    ["Iron Man", "War Machine"],
    ["Spider Man", ["MJ", "Ned"]],
    ["Capitán América", "Ant Man", "Bucky"] 
]

print(universo_marvel[1][1][0]) #Esto se hace para entrar en listas anidadas
```

## Métodos

```python
numeros = [90,12,45,8,69,71,34,16]
print(max(numeros))
print(min(numeros))

#sorted y sort
heroes_ordenados = sorted(heroes) # Ordena una copia del array heroes
print(heroes_ordenados)
heroes.sort() #Ordena el array original
print(heroes)
print(heroes[::-1])#Imprime el array ordenado alfabeticamente al reves

# copy (crea una copia independiente)
heroes_copiados = heroes.copy()
print(heroes_copiados)

#reverse (le damos la vuelta a una lista modificandola)
heroes_copiados.reverse()

#count (cuenta las veces que aparece un elemento en una lista)
print(heroes.count("Hulk"))
```

```python
# append -> añade un elemento al final de la lista
movies = ["Batman"]
movies.append("Sorcerer")
print(movies)

# extend -> añade varios elementos al final de una lista
movies.extend(("The Thing", "The Evil Dead", "El Viaje de Chihiro"))
print(movies)

# insert -> Añade un nuevo elemento a un indice concreto
movies.insert(2, "El Castillo Infinito")
print(movies)

# pop -> Elimina el último elemento de una lista
movies.pop()
print(movies)
movies.pop(2)
print(movies)

# remove -> Elimina la primera aparición del elemento que le demos por argumento
movies.remove("Batman")
print(movies)

# clear -> Limpia la lista
movies.clear()
print(movies)
```
