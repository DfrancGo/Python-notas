## Funciones

```python
def saludar():
    print("Hola!")

saludar()
```

```python
def saludar_con_nombre(nombre): #Crear el hueco es el parámetro
    print(f"Hola {nombre}!")

saludar_con_nombre("Daniel") #Darle un valor a un parámetro es el argumento
```

```python
def sumar_dos_numeros(numeroA, numeroB):
    total = numeroA+numeroB
    print(total)

sumar_dos_numeros(4,8)
```

```python
def anuncio_anio(year=2026): #Se suele dejar un valor predeterminado al parametro y luego alejecutarse la función si se le da un parametro, lo cambia
    print(f"Estamos en el año: {year}")

anuncio_anio(2023)
anuncio_anio(2025)
anuncio_anio()
```

```python
def sumar (numeroA, numeroB):
    """suma dos números y retorna el resultado

    Args:
        numeroA (int): Primer número a sumar
        numeroB (int): Segundo número a sumar

    Returns:
        int: El resultado de los dos números sumados
    """
    total = numeroA+numeroB
    return total

resultado = sumar(7,3)
print(resultado)

sumar(5,6)
```
