## Funciones Built-in
Son funciones que ya vienen por defecto en el lenguaje, sin necesidad de algún framework

```python
# print -> muestra por consola un string
print("Hola qué tal")
print("Tengo ", 2026-1999, "años")
print(f"Tengo {2026-1999} años")
```

```python
# len -> cuenta elementos/longitudes
print(len("Python"))
print(len([1,2,3,4,5,6,7,8,9]))
```

```python
# type -> consulta el tipo de dato ->class"float"
print(type("hola"))
print(type(25))
print(type(50.5))
print(type(True))
print(type(None))
print(type([1,2,3,4]))
```

```python
# dir -> Muestra los atributos o metodos que tiene un objeto
texto = "Python"
print(dir(texto))

# help -> Muestra la documentación de una función o método
help(len)
```

```python
# Conversión de tipo de datos:
print(type(int("1234")))
print(type(float("13.5")))
print(bool(0))
print(bool(1))
print(bool(65)) # -> True porque el unico que es false es el 0 
```

```python
# Funciones numéricas:
print(abs(-7)) #Valor absoluto 
print(max(3,4,5,23,67,43,98,34,1004))
print(min(3,4,5,23,67,43,98,34,1004))
print(round(3.14159, 3)) #-> Redondea los decimales que le digamos 
```

```python
# Funciones para listas o secuencias:
numeros = [2,4,6,8,10,13,1,5,7,3]
print(sum(numeros)) #->Suma todos los valores de la lista
print(sorted(numeros)) #-> Organiza la lista en orden ascendente
print(list(reversed(numeros))) #->invierte el orden del array
```

```python
# Funciones lógicas
print(any([False,True,False])) #->Comprueba si hay algun True dentro
print(all([True, True, True]))
print(all([True, False, True]))
print(all([False,False,False]))
```
