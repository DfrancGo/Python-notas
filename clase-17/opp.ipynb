## Programación Orientada a Objetos (OPP ó POO)
### ¿Cómo se crea un objeto?
Para empezar se crea un "molde" llamado clase
Dentro de esa clase se pueden definir metodos que le pertenezcan a ese objeto y todos los objetos creados con ese molde lo compartirán

```python
class Movie:
    """
    Representa una película con infromación básica
    """
    def __init__ (self, title, year, duration_min, genre): #Este es el constructor 
        self.title = str(title)
        self.year = int(year)
        self.duration_min = int(duration_min)
        self.genre = str(genre)

    def show_info(self):
        return(f"La película {self.title} es del género {self.genre}")

    def readable_duration(self):
        hours = self.duration_min // 60 #doble para quedarnos sólo la parte entera
        mins = self.duration_min % 60
        return(f"{hours}h - {mins}")

matrix = Movie("The Matrix", 1995,123, "Sci-fi")
print(matrix.year)
print(matrix.show_info())
print(matrix.readable_duration())

arrival = Movie("Arrival", 2016, 116, "Sci-fi")
print(arrival.title)
print(arrival.readable_duration())


```

```python
class Vehiculo:
    def __init__(self, nombre, color, marca):
        self.nombre = str(nombre)
        self.color = str(color)
        self.marca = str(marca)

    def avanzar(self):
        return (f"El vehiculo {self.marca} {self.nombre} está avanzando")
    def detener(self):
        return (f"El vehiculo {self.marca} {self.nombre} está detenido")

seat_ibiza = Vehiculo("Ibiza", "Gris oscuro", "Seat")
print(seat_ibiza.avanzar())

class Avion(Vehiculo):
    def __init__(self, nombre, color, marca):
        super().__init__(nombre, color, marca) #Estamos delegando la creación del avión a la superclase vehiculo

    def volar(self):
        return(f"El avión {self.marca} {self.nombre} está volando")

boeing_737 = Avion("737", "Blanco", "Boeing")

print(boeing_737.avanzar(), boeing_737.volar())

class Coche(Vehiculo):
    def __init__(self, nombre, color, marca):
        super().__init__(nombre, color, marca)
    def derrapar(self):
        return (f"El coche {self.marca} {self.nombre} está derrapando")

seat_ibiza = Coche("Ibiza", "Gris", "Seat")
print(seat_ibiza.derrapar())
```

Como se ve en el ejemplo hay una superclase llamada vehiculo de la cuál Avion y Coche heredan las mismas caracteristicas y métodos, pero la diferencia es que a cada sub clase le puedo poner sus propias caracteristicas y métodos, Avión vuela y Coche derrapa.
