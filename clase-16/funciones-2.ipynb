### args -> argumentos posicionales(se guardan en una tupla)
### 

```python
# *args -> argumentos posicionales(se guardan en una tupla)

def mostrar_nombres(*args):
    for nombre in args:
        print(nombre)


mostrar_nombres("Daniel", "Inés", "Ale", "Sandra", "Yurany")
```

```python
# **kwargs -> Es como el args pero recibe una cantidad indefinida de claves y valor
def mostrar_user (**kwargs):
    for clave, valor in kwargs.items():
        print(f"{clave}: {valor}")

mostrar_user(nombre = "Iron Man", identidad = "Tony Stark", edad = 50)
```

```python
def ejemplo(*args, **kwargs):
    print(f"args: ", args)
    print(f"kwargs: ", kwargs)

ejemplo("Hola", "Adios", nombre="Paco", apellido="Perez")
```

```python
#Función recursiva

def cuenta_atras(numero):
    print(numero)
    if(numero>1):
        cuenta_atras(numero-1)

cuenta_atras(6)
```

```python
# Funciones lambda -> PEQUEÑAS funciones que se escriben en una sola linea -> SON ANONIMAS

def suma(a,b):
    return a+b

resultado_suma = lambda a,b: a+b

print(resultado_suma(3,5))
```
