## Tuplas

```python
superheroe = ("Spiderman", "Peter Parker", "Marvel", 1962)
print(type(superheroe))
```

```python
ejemplo = [1,2,3,4,5,6]
print(ejemplo)
ejemplo[1] = 20
print(ejemplo) #En las listas se puede modificar lo que llevan dentro, en las tuplas no

tupla_ejemplo = (1,2,3,4,5,6)
#tupla_ejemplo[1]= #20 salta un error, no se puede
print(tupla_ejemplo)
```

```python
#Formas de crear una tupla
otro_ejemplo = (5,) #La coma se llama trailing coma y es necesario para crear una tupla con un sólo elemento
print(type(otro_ejemplo))
```

```python
# Indexacion con tuplas
videojuego = ("the legend of zelda", "Nintendo", 1996)
print(videojuego[0])
```

```python
# Saltarnos la inmutabilidad
videojuego = ("the legend of zelda", "Nintendo", 1996)
print(type(videojuego))
videojuego_lista  = list(videojuego)
videojuego_lista[0] = "The Legend of Zury"

videojuego = tuple(videojuego_lista)
print(videojuego)
```

Lo que se puede hacer con las tuplas es lo mismo que con los objetos (diccionarios), con la principal diferencia de que las tuplas no se pueden editar

```python
# index -> Devuelve la posición del primer elemento del valor buscado
videojuego = ("the legend of zelda", "Nintendo", 1996)
print(videojuego.index("Nintendo"))

# count -> Cuenta cuántas veces aparece un elemento en una tupla
print(videojuego.count("Nintendo"))

# len -> Cuenta el número de elmentos dentro de una tupla
print(len(videojuego))

# min y max
numeros = (1,2,3,4,5,6)
print(min(numeros))
print(max(numeros))

#
```
