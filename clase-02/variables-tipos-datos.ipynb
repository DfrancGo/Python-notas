## Variables y tipos de datos

```python
# Esto es un comentario de Python
tecnologia = "Python"

print("La variable tecnología tiene como valor: ", tecnologia)
print(f"La variable tecnología tiene como valor: {tecnologia}")
```

```python
# Tipos de datos:
# Variables siempre minusculas, barras bajas (NO CAMELCASE)

#Numbers / números
numero_entero = 50
print(type(numero_entero)) #método type() es para dar el tipo de dato
numero_decimal = 5.5
print(type(numero_decimal))
numero_complejo = 2+3j
print(type(numero_complejo))
```

```python
#String
nombre = "Daniel"
perro = "Trufa"
multilinea = """Hola, soy
un texto
multilinea"""
print(multilinea)
print(type(nombre))
```

```python
#Booleans
mayor_edad = True #Ambos valores booleanos se escriben con la primera letra en maúscula
trabaja_tec = False
print(type(mayor_edad))
```

```python
#Tipo de dato None (null en JS y Java)
tipo_dato_none = None
print(type(tipo_dato_none))
```
